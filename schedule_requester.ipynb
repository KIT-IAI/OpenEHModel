```python
def on_connect(client, userdata, flags, rc, properties=None):
    """
    Callback for when the client receives a CONNACK response from the server.
    (rc is the connection result)
    """
    if rc == 0:
        print(f"Successfully connected to MQTT Broker: {MQTT_BROKER_HOST}")
        # Subscribe to the specific topic upon successful connection
        # Using QoS 1 for at-least-once delivery
        client.publish(MQTT_TOPIC_TO_SUBSCRIBE, 'testtls', qos=0, retain=False)
        print(f"Message published to topic '{MQTT_TOPIC_TO_SUBSCRIBE}'")
    else:
        print(f"Failed to connect, return code: {rc}\nCheck broker address, port, and network.")
        # Optionally, you could try to reconnect here or exit
        # exit(1) # Or handle reconnection logic
```

```python
client = mqtt.Client(mqtt.CallbackAPIVersion.VERSION2, client_id="kit2")
client.username_pw_set(USERNAME, PASSWORD)
client.tls_set(ca_certs=None, certfile=None, keyfile=None, cert_reqs=ssl.CERT_NONE, tls_version=ssl.PROTOCOL_TLSv1_2, ciphers=None)
client.on_connect = on_connect
```

```python
try:
    # Connect to the broker
    client.connect(MQTT_BROKER_HOST, MQTT_BROKER_PORT, MQTT_KEEPALIVE_INTERVAL)
except ConnectionRefusedError:
    print(f"Connection refused. Is the broker running at {MQTT_BROKER_HOST}:{MQTT_BROKER_PORT}?")
    exit(1)
except Exception as e:
    print(f"Could not connect to MQTT broker: {e}")
    exit(1)
```
